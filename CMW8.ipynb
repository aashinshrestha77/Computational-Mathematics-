```python
import numpy as np

import matplotlib.pyplot as plt

""" Task 1: Solve linear systems graphically. """

# Generate 100 equidistant point between -5 to 5 for x

x = np.linspace(-5,5,100)

## Calculate y values for Equation 1 and 2

y1 = (3-2*x)/3

y2 = (6-4*x)/6

y3 = (1+3*x)/2

## Generate a figure frame for plotting the lines

fig, ax = plt.subplots()

## Plot the two lines

ax.plot(x,y1, color='r')

ax.plot(x,y2, color='g')

ax.plot(x,y3, color='b')

## To decorate the plot, we insert equations.

ax.text(0.5,0.6, '$2x+3y=3$', color='r')

ax.text(-3,0.5,'$4𝑥+6𝑦=6$', color='g')

ax.text(1.2,2, '$3𝑥−2𝑦=1$', color='b')

## Set the limits of the X & Y axes.

ax.set_xlim(-4,4)

ax.set_ylim(-2,6)

## Set property of X and Y axes, e.g., color and linewidth

ax.axvline(color='k',linewidth = 1)

ax.axhline(color='k',linewidth = 1)

## This option specifies the ticks based the list of numbers provided.

ax.set_xticks(list(range(-4,5)))

ax.set_aspect('equal')

ax.grid(True,ls=':')

plt.xlabel("X")

plt.ylabel("Y")

plt.show()
print (" Two lines are overlapping and one line intersects the other two at a point.")
```

1. Create a matrix 𝑨 of order 5 × 5 with random entries. Calculate the rank of the matrix 𝑨 to check the number of linearly independent columns or rows and to check if the matrix is invertible. What are the other ways to check if the matrix is invertible?

```python
import numpy as np
A = np.random.randint(1,50,(5,5))
print(A)

#rank
rank = np.linalg.matrix_rank(A)
print("Rank of the matrix:", rank)

#Check invertibility
if rank==5:
    print("The matrix is invertible")
else:
    print("The matrix is not invertible")
```

2. Generate a matrix 𝑨 of order 5 × 5 with random entries and generate a vector (column) 𝒙 of length 5 with random entries. Calculate Matrix-vector product 𝒃 = 𝑨𝒙. Now, we have a linear system, i.e. the matrix 𝑨 and a vector 𝒃. Solve this Linear System.

```python
import numpy as np
A = np.random.randint(1,50,(5,5))
x = np.random.randint(1,50,(5,1))
b = np.dot(A,x)

print("Matrix A:\n", A)
print("Vector x:\n",x)
print("Matrix-Vector Product b=Ax:",b)

try:
  x_solution = np.linalg.solve(A,b)
  print("Solution x:", x_solution)
except:
  print("Matrix is not invertible.")
```

3. Solve the following system.

8x+3y−2z=9

−4x+7y+5z==15

3x+4y−12z=35

```python
import numpy as np

A = np.array([
    [8, 3, -2],
    [-4, 7, 5],
    [3, 4, -12]
])

B = np.array([[9],
              [15],
              [35]])

X = np.linalg.solve(A, B)

print("Solution (x, y, z):\n", X)
```

1. Consider the following 10 points in X-Y plane: (−3, −7.6), (−2,1), (−1, 4.8), (0, 5), (1, 2.8), (2, −0.6), (3, −4), (4, −6.2), (5, −6), (7, 6.4). Plot these points in X-Y plane. Connect the points to guess the polynomial. Once you guess your polynomial, formulate a Linear System and then solve it to find the coefficient of the polynomial. Finally, check if the polynomial fits along with the above given points.

```python
import numpy as np
import matplotlib.pyplot as plt

# Given points
x = np.array([-3, -2, -1, 0, 1, 2, 3, 4, 5, 7])
y = np.array([-7.6, 1, 4.8, 5, 2.8, -0.6, -4, -6.2, -6, 6.4])

# Step 1: Plot given points
plt.scatter(x, y, color='black', label='Given Points')

# Step 2: Fit a cubic polynomial (best guess)
coeffs = np.polyfit(x, y, 3)   # degree 3 polynomial
a, b, c, d = coeffs

print("Polynomial coefficients:")
print(f"a = {a}, b = {b}, c = {c}, d = {d}")

# Step 3: Create polynomial function
x_smooth = np.linspace(min(x), max(x), 200)
y_smooth = np.polyval(coeffs, x_smooth)

# Step 4: Plot fitted polynomial
plt.plot(x_smooth, y_smooth, color='blue', label='Fitted Polynomial')

# Step 5: Check fit (predicted vs actual)
y_pred = np.polyval(coeffs, x)
print("\nComparison (Actual vs Predicted):")
for i in range(len(x)):
    print(f"x = {x[i]:>2}, Actual = {y[i]:>6}, Predicted = {y_pred[i]:>8.3f}")

# Final plot settings
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)
plt.title("Given Points and Fitted Polynomial")
plt.show()
```
